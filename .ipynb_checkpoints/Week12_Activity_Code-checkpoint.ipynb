```python
"""
Week 12 - Activity 1: Predict Students' Dropout and Academic Success
=====================================================================
Dataset: UCI ML Repository #697
https://archive.ics.uci.edu/dataset/697/predict+students+dropout+and+academic+success

Author: Dilini Karunarathne
Description:
    End-to-end pipeline -> data cleaning -> EDA -> preprocessing ->
    model training (Logistic Regression, Random Forest, XGBoost) ->
    evaluation -> feature importance.

How to get the data (pick ONE option):
Option A (recommended, no manual download needed):
    pip install ucimlrepo
    -> the script below fetches the dataset automatically.

Option B (manual):
    1. Go to the dataset link above -> "Download" button.
    2. Unzip -> you'll get a file called `data.csv` (semicolon-separated).
    3. Place it in the same folder as this script and set:
         USE_UCIMLREPO = False
       and update CSV_PATH below.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay
)

# XGBoost is optional - comment out if not installed (pip install xgboost)
try:
    from xgboost import XGBClassifier
    HAS_XGB = True
except ImportError:
    HAS_XGB = False

RANDOM_STATE = 42

# -------------------------------------------------------------------
# STEP 1: LOAD DATA
# -------------------------------------------------------------------
USE_UCIMLREPO = True
CSV_PATH = "data.csv"   # only used if USE_UCIMLREPO = False

if USE_UCIMLREPO:
    from ucimlrepo import fetch_ucirepo
    dataset = fetch_ucirepo(id=697)
    X_raw = dataset.data.features
    y_raw = dataset.data.targets
    df = pd.concat([X_raw, y_raw], axis=1)
else:
    df = pd.read_csv(CSV_PATH, sep=";")

print("Shape of dataset:", df.shape)
print(df.head())

# -------------------------------------------------------------------
# STEP 2: DATA CLEANING
# -------------------------------------------------------------------
# 2.1 Clean column names (remove spaces / weird chars)
df.columns = (
    df.columns.str.strip()
    .str.replace(" ", "_")
    .str.replace(r"[()/']", "", regex=True)
)

# 2.2 Check for missing values
print("\nMissing values per column:\n", df.isnull().sum()[df.isnull().sum() > 0])

# 2.3 Check for duplicate rows
dup_count = df.duplicated().sum()
print(f"\nDuplicate rows: {dup_count}")
df = df.drop_duplicates()

# 2.4 Fix data types (most UCI columns are already numeric/coded;
#     Target column is the only text/categorical one)
target_col = "Target"  # values: 'Dropout', 'Enrolled', 'Graduate'

# 2.5 If any missing values exist, impute (median for numeric)
num_cols = df.select_dtypes(include=[np.number]).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

print("\nAfter cleaning, shape:", df.shape)

# -------------------------------------------------------------------
# STEP 3: EXPLORATORY DATA ANALYSIS (EDA)
# -------------------------------------------------------------------
plt.figure(figsize=(6, 4))
sns.countplot(x=target_col, data=df, order=df[target_col].value_counts().index)
plt.title("Target Class Distribution")
plt.tight_layout()
plt.savefig("target_distribution.png")
plt.close()

print("\nTarget value counts:\n", df[target_col].value_counts())

# Correlation heatmap (numeric features only, top correlated with target
# after encoding, just for a general view here)
plt.figure(figsize=(14, 10))
sns.heatmap(df[num_cols].corr(), cmap="coolwarm", center=0)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png")
plt.close()

# -------------------------------------------------------------------
# STEP 4: PREPROCESSING
# -------------------------------------------------------------------
le = LabelEncoder()
df["Target_encoded"] = le.fit_transform(df[target_col])
print("\nTarget classes:", dict(zip(le.classes_, le.transform(le.classes_))))

X = df.drop(columns=[target_col, "Target_encoded"])
y = df["Target_encoded"]

# Train-test split (stratified because classes are imbalanced)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

# Scale features (helps Logistic Regression converge / perform better)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -------------------------------------------------------------------
# STEP 5: MODEL TRAINING
# -------------------------------------------------------------------
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE),
}
if HAS_XGB:
    models["XGBoost"] = XGBClassifier(
        n_estimators=300, use_label_encoder=False,
        eval_metric="mlogloss", random_state=RANDOM_STATE
    )

results = {}

for name, model in models.items():
    if name == "Logistic Regression":
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
    else:
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average="weighted")
    results[name] = {"accuracy": acc, "f1_weighted": f1, "model": model, "y_pred": y_pred}

    print(f"\n=== {name} ===")
    print(f"Accuracy: {acc:.4f} | Weighted F1: {f1:.4f}")
    print(classification_report(y_test, y_pred, target_names=le.classes_))

# -------------------------------------------------------------------
# STEP 6: PICK BEST MODEL + CONFUSION MATRIX
# -------------------------------------------------------------------
best_name = max(results, key=lambda k: results[k]["f1_weighted"])
print(f"\nBest model: {best_name} (F1 = {results[best_name]['f1_weighted']:.4f})")

cm = confusion_matrix(y_test, results[best_name]["y_pred"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=le.classes_)
disp.plot(cmap="Blues", xticks_rotation=45)
plt.title(f"Confusion Matrix - {best_name}")
plt.tight_layout()
plt.savefig("confusion_matrix.png")
plt.close()

# -------------------------------------------------------------------
# STEP 7: FEATURE IMPORTANCE (for tree-based best model)
# -------------------------------------------------------------------
best_model = results[best_name]["model"]
if hasattr(best_model, "feature_importances_"):
    importances = pd.Series(best_model.feature_importances_, index=X.columns)
    importances = importances.sort_values(ascending=False).head(15)

    plt.figure(figsize=(8, 6))
    sns.barplot(x=importances.values, y=importances.index)
    plt.title(f"Top 15 Feature Importances - {best_name}")
    plt.tight_layout()
    plt.savefig("feature_importance.png")
    plt.close()

    print("\nTop 10 important features:\n", importances.head(10))

print("\nDone. Charts saved as PNG files in the working directory.")
```
